## Assignment 1 — Solution


```
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
texts = [
    "I loved the cinematography but the plot was weak.",
    "Absolutely terrible, would not recommend.",
    "It was fine, nothing memorable.",
    "A brilliant performance by the lead actor!",
    "I fell asleep halfway through.",
]
for t in texts:
    r = classifier(t)[0]
    print(f"{t} -> {r["label"]} ({r["score"]:.3f})")
```

## Assignment 2 — Solution


```
from transformers import pipeline

classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
headline = "Stocks rally as tech companies report record earnings"
labels = ["finance", "sports", "politics", "entertainment"]
res = classifier(headline, candidate_labels=labels)
for label, score in zip(res["labels"], res["scores"]):
    print(f"{label}: {score:.3f}")
```

## Assignment 3 — Solution


```
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
prompt = "Once upon a time"
out = generator(prompt, max_length=40, num_return_sequences=2, temperature=0.9, do_sample=True)
for i, o in enumerate(out, 1):
    print(f"--- {i} ---
{o["generated_text"]}
")
```

## Assignment 4 — Solution


```
from transformers import pipeline

ner = pipeline("ner", aggregation_strategy="simple")
text = "Elon Musk founded Tesla in California and later acquired Twitter."
print("With aggregation:")
for e in ner(text):
    print(e["word"], e["entity_group"])

ner2 = pipeline("ner")
print("
Without aggregation:")
for e in ner2(text):
    print(e["word"], e["entity_group"])
```

## Assignment 5 — Solution


```
from transformers import pipeline

article = """
Artificial intelligence is transforming industries worldwide. From healthcare to finance,
AI systems are automating tasks and improving decision-making. However, concerns about
bias, privacy, and job displacement remain. Researchers are working on responsible AI
frameworks to ensure fair and transparent systems.
"""
summarizer = pipeline("summarization")
summary = summarizer(article, max_length=40, min_length=10, do_sample=False)[0]["summary_text"]
print("Summary:", summary)
translator = pipeline("translation", model="Helsinki-NLP/opus-mt-en-es")
print("Spanish:", translator(summary)[0]["translation_text"])
```
